# Task 4: Named Entity Recognition Exploration

In this notebook, we explore Named Entity Recognition using `spaCy`. We compare a rule-based extraction technique and two pre-trained neural NER models from `spaCy` (`en_core_web_sm` vs. `en_core_web_md`).

Finally, we visualize the entities using `displaCy`.

```python
import spacy
from spacy import displacy
import pandas as pd
import re
```

## 1. Load spaCy Models
We load two models of different sizes to compare their capabilities.

```python
try:
    nlp_sm = spacy.load('en_core_web_sm')
except OSError:
    print("Downloading en_core_web_sm...")
    spacy.cli.download('en_core_web_sm')
    nlp_sm = spacy.load('en_core_web_sm')

try:
    nlp_md = spacy.load('en_core_web_md')
except OSError:
    print("Downloading en_core_web_md...")
    spacy.cli.download('en_core_web_md')
    nlp_md = spacy.load('en_core_web_md')
```

## 2. Rule-Based Extraction
A simple rule-based approach using regex to find matching consecutive capitalized words.

```python
def extract_entities_rule_based(text):
    pattern = r'\b([A-Z][a-z]+(?:\s+[A-Z][a-z]+)*)\b'
    matches = re.finditer(pattern, text)
    
    entities = []
    stop_words = {"The", "A", "An", "In", "On", "Of", "And", "To", "With", "For", "By", "At"}
    for match in matches:
        entity_text = match.group(1)
        if entity_text not in stop_words:
            entities.append(entity_text)
    return entities

text = "Apple Inc. is considering buying a startup in London for $1 billion."
print("Rule-Based extraction:\n", extract_entities_rule_based(text))
```

## 3. Model-Based Extraction
Using spaCy's pre-trained pipeline for identifying complex entities seamlessly.

```python
doc_sm = nlp_sm(text)
print("Entities via en_core_web_sm:")
for ent in doc_sm.ents:
    print(f"{ent.text} -> {ent.label_}")
```

```python
doc_md = nlp_md(text)
print("Entities via en_core_web_md:")
for ent in doc_md.ents:
    print(f"{ent.text} -> {ent.label_}")
```

## 4. Visualizing Entities with displaCy (Bonus)

```python
# Render visualization inline if possible
displacy.render(doc_md, style="ent", jupyter=True)
```
